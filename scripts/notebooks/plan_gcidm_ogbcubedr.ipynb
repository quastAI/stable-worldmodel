# Plan with GCIDM on OGBCubeDR (RunPod)

Runs `scripts/plan/eval_wm.py` against a trained GCIDM checkpoint on
`cube_quadruple_dr_expert`, in **three arms** that separate the two things
GCIDM claims:

| Arm | Config | Solver | Isolates |
|---|---|---|---|
| **A** | `cube_quadruple_dr` | `CEMSolver` | what the *objective* bought (vs. the SMWM/LeWM baselines, same solver) |
| **B** | `cube_quadruple_dr_gcidm` | `PolicyCEMSolver` | what the *warm start* bought (vs. arm A, same checkpoint) |
| **C** | `cube_quadruple_dr` | `CEMSolver` | the SMWM baseline, for the A-vs-C comparison |

Running A and B off the **same checkpoint** is the point: a single number
cannot tell a better representation apart from a better initialization, and
these two arms differ only in the solver.

**Arm A needs no GCIDM-specific flags.** Planning needs only the `Dynamics`
surface (`encode`/`rollout`), which GCIDM inherits from LeWM unchanged; cost
comes from `planning.ShootingCostEvaluator` + `GoalMSE`, not from the model.
`load_pretrained` reads the checkpoint's `config.json`, so the policy head is
restored either way — in arm A it simply goes unused.

**Arm B** swaps in `PolicyCEMSolver`, which asks the model for
`get_action_distribution(info, horizon)` and uses the returned `(mu, sigma)`
as CEM's iteration-0 mean and spread instead of `N(0, var_scale)`. Iterations
1..n are ordinary CEM. `cube_quadruple_dr_gcidm.yaml` also sets
`plan_config.warm_start: false`, so the previous plan's tail is not carried
over and the head really owns the initialization.

Three settings must agree or the run is meaningless — `PolicyCEMSolver`
raises rather than silently padding:

```
plan_config.horizon (5) == wm.goal_horizon used at training time
plan_config.horizon * plan_config.action_block (25) == eval.goal_offset_steps
plan_config.history_len (3) == wm.history_size
```

Set `BASELINE_RESULTS_DIR` in the config cell to put the SMWM/LeWM baseline
numbers side by side with these in §9.

Prerequisite: a trained checkpoint under
`$STABLEWM_HOME/checkpoints/<OUTPUT_MODEL_NAME>/` (e.g. from
`train_gcidm_ogbcubedr.ipynb` on this same volume) and the
`cube_quadruple_dr_expert` dataset under `$STABLEWM_HOME/datasets/`.

`load_pretrained` needs one specific `.pt` file — it errors on an ambiguous folder with several. Auto-picks the latest epoch on this volume unless `POLICY_EPOCH` is set above.

````python
import glob
import re

ckpt_files = glob.glob(os.path.join(CHECKPOINT_DIR, 'weights_epoch_*.pt'))
assert ckpt_files, f'No weights_epoch_*.pt found in {CHECKPOINT_DIR}'

epochs = sorted(int(re.search(r'weights_epoch_(\d+)\.pt$', f).group(1)) for f in ckpt_files)
epoch = POLICY_EPOCH if POLICY_EPOCH is not None else epochs[-1]
assert epoch in epochs, f'weights_epoch_{epoch}.pt not found; available epochs: {epochs}'

POLICY = f'{OUTPUT_MODEL_NAME}/weights_epoch_{epoch}.pt'
print('Available epochs:', epochs)
print('Using policy     :', POLICY)
````

### Arm B — the same checkpoint, CEM warm-started from the policy head

Only the solver changes. Any difference against arm A is attributable to the
warm start alone.

`solver.var_floor` is the safety valve: it lower-bounds the proposal std so a
confidently-wrong head cannot collapse the search onto its own plan. Raise it
if arm B *underperforms* arm A — that is the signature of an over-confident
`sigma` starving CEM of exploration.

````python
!python scripts/plan/eval_wm.py --config-name=cube_quadruple_dr_gcidm \
    policy="$POLICY" \
    eval.num_eval=$NUM_EVAL_EPISODES \
    output.filename=warmstart_full_task_results.txt
````

#### Is the warm start actually doing work?

A better final success rate is the outcome that matters, but it does not prove
the head helped: CEM converges on its own, so a useless proposal that gets
overwritten looks the same as no proposal. The direct evidence is the cost at
**iteration 0** — with a good warm start it starts lower. `n_steps=1` isolates
that: it gives CEM exactly one iteration, so the reported cost *is* the
proposal's cost.

If A and B tie at `n_steps=30` but B wins at `n_steps=1`, the head is right and
CEM was already finding the same plan unaided — which is itself the useful
finding, and the argument for cutting `n_steps` (the whole point of a warm
start is a cheaper planner).

````python
for arm, cfg_name in [('cold', 'cube_quadruple_dr'),
                      ('warm', 'cube_quadruple_dr_gcidm')]:
    print(f'=== {arm} start, 1 CEM iteration ===')
    !python scripts/plan/eval_wm.py --config-name={cfg_name} \
        policy="$POLICY" \
        eval.num_eval=8 \
        solver.n_steps=1 \
        output.filename=oneshot_{arm}_results.txt
````

## 9. Compare results

Arm A (`full task`) vs the baseline dir answers *did the objective help?*
Arm B (`full task + policy warm start`) vs arm A answers *did the warm start
help?* Keep them apart when writing the result up — they are different claims.


````python
runs = {
    'A: full task, plain CEM': full_task_dir,
    'A: original (num_subgoals=1)': original_dir,
    'B: full task + policy warm start': warmstart_dir,
}
if RUN_RANDOM_BASELINE:
    runs['random baseline'] = random_dir
if BASELINE_RESULTS_DIR:
    assert os.path.isdir(BASELINE_RESULTS_DIR), BASELINE_RESULTS_DIR
    runs['C: SMWM/LeWM baseline'] = BASELINE_RESULTS_DIR

for label, run_dir in runs.items():
    txt_files = glob.glob(os.path.join(run_dir, '*_results.txt'))
    print(f'=== {label} ===')
    if not txt_files:
        print(f'  no results file in {run_dir}')
        print()
        continue
    text = open(txt_files[0]).read()
    results_section = text.split('==== RESULTS ====')[-1]
    print(results_section.strip())
    print()
````
